```python
import sklearn
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = ndf.drop(columns=["File","1x[Vibration]","2x[Vibration]","3x[Vibration]","0-4k_Hz[Vibration]"])
y = ndf["0-4k_Hz[Vibration]"]
```

```python
X
```

```python
y
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestRegressor
regressor = RandomForestRegressor(n_estimators=10,max_depth=4)
regressor.fit(X_train,y_train)
```

```python
regressor.score(X_train,y_train)
```

```python
regressor.score(X_test,y_test)
```

```python
y_preds = regressor.predict(X_test)
y_preds
```

```python
from sklearn.model_selection import cross_val_score,cross_val_predict,GroupKFold

gkf = GroupKFold(n_splits=5)
regressor = RandomForestRegressor(n_estimators=10,random_state=42,max_depth=4)
cross = cross_val_score(regressor,
                        X,
                        y,
                        cv=gkf,
                        scoring="r2",
                       groups=df["File"])

y_pred = cross_val_predict(regressor,
                        X,
                        y,
                        cv=gkf,
                        groups=df["File"])
cross.mean()
```
